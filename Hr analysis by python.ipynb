```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=df['YearsWithCurrManager'])
plt.show()
```

```python
median_value = df['YearsWithCurrManager'].median()

df['YearsWithCurrManager'].fillna(median_value, inplace=True)
```

```python
df['YearsWithCurrManager'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
# Check duplicate rows
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()

# Verify duplicates are removed
print("Duplicate Rows After Cleaning:", df.duplicated().sum())
```

```python
sns.countplot(x='Attrition', data=df)
plt.show()
```

```python
sns.countplot(x='Department',
              hue='Attrition',
              data=df)

plt.xticks(rotation=45)
plt.show()
```

```python
numeric_cols = df.select_dtypes(include=np.number)

corr = numeric_cols.corr()

plt.figure(figsize=(12,8))
sns.heatmap(corr)
plt.show()
```

```python
df['SalarySlab'] = pd.cut(
    df['MonthlyIncome'],
    bins=[0,5000,10000,15000,50000],
    labels=['Low','Medium','High','Very High']
)
```

```python
df.SalarySlab
```

```python
df['AgeBand'] = pd.cut(
    df['Age'],
    bins=[18,25,35,45,60],
    labels=['18-25','26-35','36-45','46-60']
)
```

```python
df.AgeBand
```

```python
df.to_csv("HR_Cleaned.csv",index=False)
```
